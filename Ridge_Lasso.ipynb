```python
class LassoGD:

    def __init__(self, lr, epochs, alpha):
        self.lr = lr
        self.epochs = epochs
        self.alpha = alpha

        self.m = 100
        self.b = -120

    def fit(self, X, y):

        X = X.ravel()

        for i in range(self.epochs):

            y_pred = self.m * X + self.b
            error = y - y_pred

            loss_slope_b = -2 * np.sum(error)

            loss_slope_m = (
                -2 * np.sum(error * X)
                + self.alpha * np.sign(self.m)
            )

            self.m -= self.lr * loss_slope_m
            self.b -= self.lr * loss_slope_b

    def predict(self, X):
        return self.m * X + self.b
```

```python
class RidgeGD:

    def __init__(self, lr, epochs, alpha):
        self.lr = lr
        self.epochs = epochs
        self.alpha = alpha

        self.m = 100
        self.b = -120

    def fit(self, X, y):

        X = X.ravel()

        for i in range(self.epochs):

            y_pred = self.m * X + self.b
            error = y - y_pred




            loss_slope_b = -2 * np.sum(error)

            loss_slope_m = (
                -2 * np.sum(error * X)
                + 2 * self.alpha * self.m
            )

            self.m -= self.lr * loss_slope_m
            self.b -= self.lr * loss_slope_b

    def predict(self, X):
        return self.m * X + self.b
```

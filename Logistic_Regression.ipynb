```python
import numpy as np 
```

```python
class LogisticRegression:

    def __init__(self,epoches,lr,w,b=0):
        self.w=w
        self.b=b
        self.epoches=epoches
        self.lr=lr

    def sigmoid(self,z):
        return 1/(1+np.exp(-z))
    
    def fit(self,X,y):
        m, n = X.shape
        self.w=np.zero(n)
        for _ in range(self.epoches):
                z=np.dot(self.w,X)+self.b
                y_pred=self.sigmoid(z)

                loss_w=(1/m)*(np.dot(X.T,(y_pred-y)))
                loss_b=(1/m)*(np.sum(y_pred-y))

                self.w=self.w-self.lr*loss_w
                self.b=self.b-self.lr*loss_b

        return self.w,self.b
```
